```python
import pandas as pd

df = pd.read_csv("Day8_Ecommerce_Sales_Dataset.csv")

df.head()
```

```python
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nColumns:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)
```

```python
# Select Product column
print(df["Product"])

# Select Product and Total Sales
print(df[["Product", "Total_Sales"]])
```

```python
high_sales = df[df["Total_Sales"] > 5000]

print(high_sales)
```

```python
print(df[df["Quantity"] > 5])
```

```python
print(df[df["Payment_Method"] == "UPI"])
```

```python
df_sorted = df.sort_values("Total_Sales", ascending=False)

print(df_sorted.head(10))
```

```python
total_sales = df["Total_Sales"].sum()

print("Total Sales:", total_sales)
```

```python
average_sales = df["Total_Sales"].mean()

print("Average Sales:", average_sales)
```

```python
print("Highest Sale:", df["Total_Sales"].max())
print("Lowest Sale:", df["Total_Sales"].min())
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
number_of_orders = df["Order_ID"].nunique()

print("Number of Orders:", number_of_orders)
```

```python
category_sales = df.groupby("Category")["Total_Sales"].sum()

print(category_sales)
```

```python
category_quantity = df.groupby("Category")["Quantity"].sum()

print(category_quantity)
```

```python
city_sales = df.groupby("City")["Total_Sales"].sum()

print(city_sales)
```

```python
payment_sales = df.groupby("Payment_Method")["Total_Sales"].sum()

print(payment_sales)
```

```python
average_category_sales = df.groupby("Category")["Total_Sales"].mean()

print(average_category_sales)
```

```python
top_products = (
    df.groupby("Product")["Total_Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(5)
)

print(top_products)
```

```python
top_orders = df.sort_values(
    "Total_Sales",
    ascending=False
).head(5)

print(top_orders[["Order_ID", "Product", "City", "Total_Sales"]])
```

```python
highest_rated = df.sort_values(
    "Rating",
    ascending=False
).head(5)

print(highest_rated[["Product", "Rating"]])
```

```python
category_analysis = df.groupby("Category").agg(
    Total_Sales=("Total_Sales", "sum"),
    Average_Sales=("Total_Sales", "mean"),
    Total_Quantity=("Quantity", "sum"),
    Number_of_Orders=("Order_ID", "count")
)

print(category_analysis)
```
